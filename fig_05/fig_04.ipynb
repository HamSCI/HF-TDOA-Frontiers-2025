```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
import os
```

```python
plt.rcParams['font.size']      = 18
plt.rcParams['font.weight']    = 'bold'
plt.rcParams['axes.titleweight']    = 'bold'
plt.rcParams['axes.grid']      = True
plt.rcParams['axes.xmargin']   = 0
plt.rcParams['grid.linestyle'] = ':'
```

```python
y_min = -35000   
y_max = 35000    

file_path = os.path.join('wav','40M-EM12jw-152239Z-8April24-EL90nn.wav')

sample_rate, data = wavfile.read(file_path)

duration = len(data) / sample_rate
time = np.linspace(0, duration, num=len(data))

if len(data.shape) == 2:
    data = data[:, 0]

plt.figure(figsize=(12, 4))
plt.plot(time, data)
plt.title("2024 Eclipse Waveform Tx: WA5FRF Rx: N5DUP at 16:36 UTC (317 km path)")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.xlim(13.5,duration-0.5)
plt.ylim(y_min, y_max)
plt.grid(True)
plt.tight_layout()
plt.show()
```

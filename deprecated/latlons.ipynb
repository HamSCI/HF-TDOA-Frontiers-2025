```python
import hf_tdoa as tdoa
```

```python
# WA5FRF
# https://maps.app.goo.gl/8opMdL6yvuy5txAz9
tx_lat, tx_lon = (29.57489219056098, -98.88698482175624)

# AB5YO
# https://maps.app.goo.gl/X5sZBUNEEhH4VetL6
rx_lat, rx_lon = (29.57142870343933, -98.97480900586837)
```

```python
dist_km = tdoa.geopack.greatCircleDist(tx_lat,tx_lon,rx_lat,rx_lon) * 6371
```

```python
print(dist_km)
```

```python
precision = 6
tx_grid = tdoa.locator.latlon2gridsquare(tx_lat,tx_lon,precision=precision)
rx_grid = tdoa.locator.latlon2gridsquare(rx_lat,rx_lon,precision=precision)
grid_km = tdoa.locator.grid_range_km(tx_grid,rx_grid)

print(f'TX Grid: {tx_grid}')
print(f'RX Grid: {rx_grid}')
print(f'Grid Distance: {grid_km} km')
```
